# Outlier Detection Module Assesment


This notebook serves as a comprehensive guide for conducting an outlier detection experiment, specifically focusing on evaluating the performance of an outlier detection module across three distinct types of outliers: arbitrary anomalies, typographical errors, and contextual outliers. 

```python
import pandas as pd
from find_outliers import *

import numpy as np

np.random.seed(0) 
```

## Arbitrary Anomalies

This part of experiment tries to detect "complete nonsense" outliers, the data points that are entirely unrelated or absurd in the context of the dataset.

```python
df = pd.read_csv('../datasets/carprices/car_prices.csv')


random_values = [    
    "flkjerwpoijzmv.x,cn",
    "qwiubzxnmahsgdfjkl",
    "ertqwuopiadsflkjgh",
    "zmxncbvqweoiruty",
    "plmoknijbuhvygctfxrdzeswaq",
    "kjfie9021u3jdksla;",
    "zxncbvqweurtyiop",
    "qazwsxedcrfvtgbyhnujmik,ol.p;/-",
    "1234567890qwertyuiopasdfghjklzxcvbnm",
    "!@#$%^&*()_+{}|:\"<>?[];'./,`~"]

num_errors = 100
error_indices = np.random.choice(df.index, size=num_errors, replace=False)


for idx in error_indices:
    df.at[idx, 'make'] = np.random.choice(random_values)



outlier_indices = identify_categorical_outliers(df, 'make')


common_indices = list(set(error_indices) & set(outlier_indices))


rate_of_error_indices = len(common_indices) / len(error_indices)

rate_of_error_indices
```
